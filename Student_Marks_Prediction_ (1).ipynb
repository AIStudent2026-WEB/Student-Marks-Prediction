```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [20, 25, 35, 40, 50, 60, 70, 85]

print("Study Hours:", hours)
print("Student Marks:", marks)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1], [2], [3], [4], [5], [6], [7], [8]]
y = [20, 25, 35, 40, 50, 60, 70, 85]

model.fit(X, y)

print("Model Training Successful!")
```

```python
hours = 5

predicted_marks = model.predict([[hours]])

print("Study Hours:", hours)
print("Predicted Marks:", predicted_marks[0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, label="Actual Marks")

plt.plot(X, model.predict(X), label="Prediction Line")

plt.xlabel("Study Hours")
plt.ylabel("Student Marks")
plt.title("Student Marks Prediction using ML")

plt.legend()
plt.show()
```
